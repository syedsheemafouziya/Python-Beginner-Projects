```python
""" Mini Project: Emoji Converter

Convert text-based emotions into emojis.
# Emoji Converter Basic Version (No if, no loop)

Example Run:

Enter your message Output: Hello :) I am learning Python from Saumya :D
output: Hello😊 I am learning Python😀"""
msg= input("Enter the message: ")
msg = msg.replace(":)","😊")
msg = msg.replace(":(","☹️")
msg=msg.replace(":D","😁")
msg=msg.replace(";)","😉")
msg=msg.replace(":3","😗 ")
msg=msg.replace(":/","🫤")
print(msg)
```
