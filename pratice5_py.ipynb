```python
#Write a Python program to print numbers from 1 to 10 using a while loop.
i=1
while i <= 10:
  print(i)
  i+=1
```

```python
"""Write a program to print numbers from 10 down to 1 using a while loop.
(Hint: start from 10 and decrease the counter each time.)
"""
i=10
while i >= 1:
  print(i)
  i-=1
```

```python
"""
Write a program to print all even numbers between 1 and 50 using a while
loop.
(Hint: Use the modulus operator % to check for even numbers.)
Example Output: 2 4 6 8 ... 50
"""
i=1
while i <= 50:
  if i%2==0:
    print(i)
  i+=1
```

```python
"""
Write a program that prints the sum of first n natural numbers.
For example, if n = 5, then output should be 1 + 2 + 3 + 4 + 5 = 15.
(Hint: Keep a running total inside the loop.)
"""
sum=0
i=1
n=int(input("enter a number: "))
while i<=n:
   sum =sum+i
   i+=1
print(sum)



```

```python
"""
Write a program to print this pattern using a while loop:
*
* *
* * *
* * * *
(Hint: use one while loop and string multiplication like '*' * count)
"""
i=1
while i<=5:
  print(i*'*')
  i+=1
```

```python
"""
Saumya wants to print her name 5 times, but each time with a number in
front of it. Write a program using a while loop that prints:
1. Saumya Singh
2. Saumya Singh
3. Saumya Singh
4. Saumya Singh
5. Saumya Singh"""
i=1
while i<=5:
  print(f"{i}. saumya singh")
  i+=1
```

```python
"""
Write a program to print the multiplication table of any number using a while
loop.
(Hint: Start i = 1 and run the loop until i <= 10.)
Example Output:
Enter a number: 3
3 x 1 = 3
3 x 2 = 6
...
3 x 10 = 30"""
i=1
n=int(input("enter a value: "))
while i<=10:
  print(f"{n} * {i} = {n*i}")
  i+=1
```
