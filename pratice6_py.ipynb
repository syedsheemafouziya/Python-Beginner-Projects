```python
"""Write a program using for and range() to print all even numbers between 1
and 20."""
for i in range(2,21,2):
  print(i)
```

```python
"""
Write a program to print numbers from 1 to 50, but print "sheema fouziya"
instead of numbers that are multiples of 5.
Example Output: 1 2 3 4 sheema fouziya 6 7 8 9 sheema fouziya ..."""
for i in range(1,51):

  if i%5==0:
    print("sheema fouziya")
  else:
    print(i)
```

```python
"""
Write a program to print the square of each number from 1 to 10 using a for
loop.
Example Output: 1 4 9 16 25 36 49 64 81 100"""
for i in range(1,11):
  print(i**2)
```

```python
"""Write a program that prints the multiplication table of any number entered by
the user using a for loop."""
n=int(input("enter a number: "))
for i in range(1,11):
  print(f"{n} * {i} = {n*i}")
```

```python
#Write a program that prints all numbers from 100 to 1 using for and range().
for i in range(100,0,-1):
  print(i)
```

```python
"""You are given a list of Sheema's favorite foods. Write a Python program to
print each food item using a for loop."""
foods = ["briyani","cold coffe","carmel popcorn","chicken","wrap","pizza"]
for food in foods:
  print(food)
```

```python
"""Saumya has created a tuple of countries she has already traveled to. Write a
Python program to print each country using a for loop.
countries_traveled = ("Malaysia", "Vietnam", "Switzerland", "Italy", "Bhutan")"""
countries_traveled = ("Malaysia", "Vietnam", "Switzerland", "Italy", "Bhutan")
for country in countries_traveled:
  print(country)
```

```python
"""Write a program that prints numbers 1 to 10, but skips the number 7 using the
continue statement."""
for i in range(1,11):
  if i == 7:
    continue
  print(i)
```

```python
"""Write a program using nested loops to print this pattern:
*
* *
* * *
"""
for i in range(1,2):
  for j in range(1,4):
      print(i*j*"*")
```
