```python
"""Write a Python program that takes a number as input and prints:

"Positive" if number > 0

"Zero" if number == 0

"Negative" if number < 0"""
num=int(input("enter a number"))
if(num==0):
  print("Zero")
elif(num>0):
  print("positive")
else:
  print("negative")
```
